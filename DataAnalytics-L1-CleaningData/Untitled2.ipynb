```python
# Oasis Infobyte - Data Analytics
# Level 1 - Task 3: Cleaning Data

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
# Load the dataset

url = "https://raw.githubusercontent.com/plotly/datasets/master/diabetes.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Shape:", df.shape)

display(df.head())
```

```python
# Step 3: Data Quality Report

print("Dataset Shape:", df.shape)

print("\n--- Missing Values ---")
print(df.isnull().sum())

print("\n--- Duplicate Rows ---")
print("Duplicate count:", df.duplicated().sum())

print("\n--- Data Types ---")
print(df.dtypes)

print("\n--- Descriptive Statistics ---")
display(df.describe())

print("\n--- Unique Values ---")
print(df.nunique())
```

```python
# Step 4: Handle Missing Values

missing_before = df.isnull().sum()

print("Missing values before cleaning:")
display(missing_before[missing_before > 0])

# Replace missing values with the median of each column
numeric_columns = df.select_dtypes(include=np.number).columns

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())

print("\nMissing values after cleaning:")
display(df.isnull().sum())

print("\nMissing values handled successfully!")
```

```python
# Step 5: Remove Duplicate Rows

duplicates_before = df.duplicated().sum()

print("Duplicate rows before cleaning:", duplicates_before)

df = df.drop_duplicates()

duplicates_after = df.duplicated().sum()

print("Duplicate rows after cleaning:", duplicates_after)
print("Duplicate rows removed:", duplicates_before - duplicates_after)
```

```python
# Step 6: Check for inconsistent values

print("Unique values in each column:")
for col in df.columns:
    print(f"\n{col}:")
    print(df[col].unique()[:20])

print("\nCurrent data types:")
print(df.dtypes)
```

```python
# Step 7: Detect outliers using IQR

numeric_cols = df.select_dtypes(include=np.number).columns

outlier_summary = {}

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ((df[col] < lower) | (df[col] > upper)).sum()
    outlier_summary[col] = outliers

outlier_summary_df = pd.DataFrame(
    list(outlier_summary.items()),
    columns=["Column", "Outlier_Count"]
)

outlier_summary_df
```

```python
# Step 8: Handle invalid zero values

invalid_cols = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI"
]

# Convert invalid zeros to NaN
df[invalid_cols] = df[invalid_cols].replace(0, np.nan)

print("Missing values after replacing invalid zeros:")
print(df.isnull().sum())

# Fill missing values with median
for col in invalid_cols:
    df[col] = df[col].fillna(df[col].median())

print("\nMissing values after median imputation:")
print(df.isnull().sum())
```

```python
# Step 9: Before vs After data quality summary

summary = pd.DataFrame({
    "Column": df.columns,
    "Data_Type": df.dtypes.astype(str).values,
    "Missing_Values": df.isnull().sum().values,
    "Unique_Values": df.nunique().values
})

summary
```

```python
# Step 10: Save cleaned dataset

df.to_csv("cleaned_diabetes_dataset.csv", index=False)

print("Cleaned dataset saved successfully!")
```

```python
# Step 11: Check and remove duplicate records

duplicate_count = df.duplicated().sum()

print("Duplicate rows found:", duplicate_count)

df = df.drop_duplicates().reset_index(drop=True)

print("Dataset shape after removing duplicates:", df.shape)
```

```python
# Step 12: Save final cleaned dataset

df.to_csv("cleaned_diabetes_dataset.csv", index=False)

print("Final cleaned dataset saved successfully!")
print("Final shape:", df.shape)
```
